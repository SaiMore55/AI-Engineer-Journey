Exercise 1 – Missing values

Create a DataFrame with 10 records and missing values in two columns. Display the missing-value count and percentage for each column

```python
data = {
    "Name" : ["Sai","Kiran","Rupesh","Dipak","Viraj"],
    "Department" : ["Computer",None,"IT",None,"Computer"],
    "Age" : [20,21,22,None,19]
}

df = pd.DataFrame(data)

missing_values = df.isnull().sum()
percentage = (missing_values / len(df) ) * 100

print(percentage)
```

Exercise 3 – Outlier detection

Given:

marks = pd.Series([
    45, 50, 55, 60, 65, 70, 75, 80, 85, 200
])

Use the IQR method to identify potential outliers.

```python
marks = pd.Series([
    45, 50, 55, 60, 65, 70, 75, 80, 85, 200
])

Q1 = marks.quantile(0.25)
Q3 = marks.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * (IQR)
upper_bound = Q3 + 1.5 * (IQR)

outliers = []

for mark in marks:
    if mark < lower_bound:
        outliers.append(mark)
    elif mark > upper_bound:
        outliers.append(mark)
print(outliers)
```
