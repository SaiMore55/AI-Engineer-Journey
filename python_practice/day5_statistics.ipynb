```python
import pandas as pd
import numpy as np

data = {
    "Student": ["Amit", "Rahul", "Priya", "Sneha",
                "Rohan", "Neha", "Karan", "Pooja"],
    "Marks": [85, 72, 95, 48, 65, 88, 55, 91]
}

df = pd.DataFrame(data)

marks = df["Marks"]

print("Mean:", marks.mean())
print("Median:", marks.median())
print("Mode:", marks.mode().tolist())

print("Range:", marks.max() - marks.min())
print("Variance:", marks.var())
print("Standard Deviation:", marks.std())

df["Z_Score"] = (
    marks - marks.mean()
) / marks.std()

print("\nStatistical Analysis:")
print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = np.random.normal(
    loc=70,
    scale=10,
    size=1000
)

sns.histplot(
    data,
    kde=True,
    bins=30
)

plt.title("Normal Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```
