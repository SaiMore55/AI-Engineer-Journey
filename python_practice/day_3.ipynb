Exercise 2 – Sorting

Create a DataFrame containing five employees with their names and salaries. Sort them from highest to lowest salary.

```python
data = {
        "Names" : ["Sai","Kiran","Viraj","Rupesh","Dipak"],
        "Salaries" : [700000,500000,600000,400000,500000]
}

employee = pd.DataFrame(data)

print(employee.sort_values(by="Salaries",ascending=False))
```

Exercise 3 – Missing values

Create a DataFrame with missing values in the Age column. Replace them with the average age.

```python
data = {
        "Names" : ["Sai","Kiran","Viraj","Rupesh","Dipak"],
        "Salaries" : [700000,None,600000,400000,None],
        "Age" : [None,20,21,None,20]
}

df = pd.DataFrame(data)
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salaries"] = df["Salaries"].fillna(df["Salaries"].mean())
print(df)
```

Exercise 4 – Grouping

Create a DataFrame with student names, departments and marks. Calculate the average marks for each department.

```python
data = {
    "Names" : ["Sai","Jitendra","Rupesh","Harshal","Lokesh","Krushna"],
    "Department" : ["Computer","Electrical","Computer","Mechanical","Electrical","Mechanical"],
    "Marks" : [98,92,93,90,78,88]
}

df = pd.DataFrame(data)

avg_marks = df.groupby("Department")["Marks"].mean()

print(avg_marks)
```
