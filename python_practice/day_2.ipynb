Exercise 1 — Array fundamentals
Create a NumPy array containing [12, 24, 36, 48, 60]. Print its dimensions, shape, size, data type and average.

```python
import numpy as np

nums = np.array([12, 24, 36, 48, 60])

print(nums.ndim)
print(nums.shape)
print(nums.size)
print(nums.dtype)
print(nums.mean())
```

Exercise 2 — Array slicing
Create an array from 1 to 20. Extract the first five elements, the last five elements, and every second element.

```python
import numpy as np

nums = np.arange(1,21)

print(nums[:5])
print(nums[-5:])
print(nums[::2])
```

Exercise 3 — Reshaping
Create an array from 1 to 24 and reshape it into 4 rows and 6 columns. Print the second row and fourth column.

```python
import numpy as np

nums = np.arange(1,25)
reshaped = nums.reshape(4,6)

print(reshaped)
print("Second row: ",reshaped[1])
print("Fourth Column: ",reshaped[:,3])
```

Exercise 4 — Vectorization
Create an array of marks [45, 60, 75, 80, 90]. Increase every student's marks by 5 using one NumPy operation.

```python
import numpy as np

marks = np.array([45, 60, 75, 80, 90])

print(marks+5)
```

Exercise 5 — Broadcasting
Create a 3x3 array containing [[10,20,30],[40,50,60],[70,80,90]]. Add [1,2,3] to every row.

```python
import numpy as np

arr = np.array([[10,20,30],[40,50,60],[70,80,90]])

print(arr+[1,2,3])
```

Exercise 6 — Matrix multiplication
Create two 2x2 matrices A=[[2,3],[4,5]] and B=[[1,2],[3,4]]. Calculate their matrix product and their element-wise product.

```python
import numpy as np

A = np.array([[2,3],[4,5]])
B = np.array([[1,2],[3,4]])

print(A @ B)

print(A * B)
```
